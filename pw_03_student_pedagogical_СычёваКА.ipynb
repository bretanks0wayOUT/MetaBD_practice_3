### Пороговый отбор по текущим оценкам

Правило: **учащийся предварительно выделяется для анализа, если последняя доступная оценка ниже порога `t`** (обычно G2; для варианта 11 — G1). В историческом наборе проверяем, насколько отбор соответствует уже известному итоговому результату: `G3 ниже 10 = 1`, если `G3 < 10`.

| Известная итоговая оценка | Не выделен правилом | Выделен правилом |
|---|---|---|
| **G3 ≥ 10** | TN — корректно не выделен | FP — выделен дополнительно, хотя G3 не ниже 10 |
| **G3 < 10** | FN — случай не выявлен правилом | TP — случай выявлен правилом |

**Полнота выявления (Recall):**

$$
\mathrm{Recall}=\frac{\mathrm{TP}}{\mathrm{TP}+\mathrm{FN}}
$$

**Точность отбора (Precision):**

$$
\mathrm{Precision}=\frac{\mathrm{TP}}{\mathrm{TP}+\mathrm{FP}}
$$

**Педагогический смысл.** При увеличении порога обычно выделяется больше учащихся, но растёт количество дополнительных случаев рассмотрения (FP). Ни TP, ни FN, ни FP **не указывают**, была ли реально проведена консультация. Решение о необходимости поддержки требует беседы и предметной диагностики. Это ретроспективная проверка правила, **не** независимое доказательство качества будущего прогноза.

### Оговорки о данных

- `failures` — количество ранее не освоенных учебных курсов (0–3), **не** личные «неудачи» учащегося.
- `goout` — частота встреч с друзьями (1–5), **не** пропуски занятий.
- `absences` — число пропущенных занятий, но данные не раскрывают причины каждого пропуска.
- `G3 = 0` при `G1 > 0` — сочетание оценок, причины которого **нельзя установить из данных**; оно не является доказательством неявки на экзамен.
- Группы F/C/B/A определены в этом упражнении по выбранным диапазонам оценок, а не обозначают официальную педагогическую диагностику.
- Зависимости между признаками статистические: по ним нельзя устанавливать причинность или оценивать способности ученика.

````python
# @title Служебная ячейка: варианты, поля и показатели DataLens (просто запустите) { display-mode: "form" }
import html
import numpy as np
import pandas as pd
from IPython.display import display, Markdown

# ---------------------------------------------------------------------------
# Показатели (вычисляемые поля датасета DataLens) и их расчёт в pandas
# ---------------------------------------------------------------------------
MEASURES = {
    "Учащихся":               ("COUNT([Код])",                                   lambda d: len(d)),
    "Учащихся с G3 ниже 10":              ("SUM([G3 ниже 10])",                                    lambda d: int(d["G3 ниже 10"].sum())),
    "Доля G3 ниже 10, %":          ("ROUND(AVG([G3 ниже 10]) * 100, 1)",                    lambda d: round(d["G3 ниже 10"].mean() * 100, 1)),
    "Средняя G1":             ("ROUND(AVG([G1]), 2)",                            lambda d: round(d["G1"].mean(), 2)),
    "Средняя G2":             ("ROUND(AVG([G2]), 2)",                            lambda d: round(d["G2"].mean(), 2)),
    "Средняя G3":             ("ROUND(AVG([G3]), 2)",                            lambda d: round(d["G3"].mean(), 2)),
    "Средние пропуски":       ("ROUND(AVG([Пропуски]), 1)",                      lambda d: round(d["Пропуски"].mean(), 1)),
    "Средняя динамика":       ("ROUND(AVG([Динамика G3-G1]), 2)",                lambda d: round(d["Динамика G3-G1"].mean(), 2)),
    "Выделено правилом":               ("SUM([Выделен правилом])",                                  lambda d: int(d["Выделен правилом"].sum())),
    "Подтверждено по G3":        ("SUM([TP])",                                      lambda d: int(d["TP"].sum())),
    "Не выявлено при G3 ниже 10":    ("SUM([FN])",                                      lambda d: int(d["FN"].sum())),
    "Полнота выявления, %":   ("IF SUM([G3 ниже 10]) > 0 THEN ROUND(100.0 * SUM([TP]) / SUM([G3 ниже 10]), 1) ELSE NULL END",
                               lambda d: round(100 * d["TP"].sum() / d["G3 ниже 10"].sum(), 1) if d["G3 ниже 10"].sum() else np.nan),
    "Точность отбора, %":   ("IF SUM([Выделен правилом]) > 0 THEN ROUND(100.0 * SUM([TP]) / SUM([Выделен правилом]), 1) ELSE NULL END",
                               lambda d: round(100 * d["TP"].sum() / d["Выделен правилом"].sum(), 1) if d["Выделен правилом"].sum() else np.nan),
}
KPI = ["Учащихся", "Доля G3 ниже 10, %", "Средняя G3", "Полнота выявления, %"]   # индикаторы — на каждом дашборде

# ---------------------------------------------------------------------------
# Типы чартов: как раскладывать поля по секциям визарда DataLens
# ---------------------------------------------------------------------------
def C(t, title, **kw):
    return dict(t=t, title=title, **kw)

S = "список, множественный выбор"
S1 = "список, один вариант"
INP = "поле ввода, операция «Больше или равно»"

FEATURES = {
    "filter": "**Фильтрация чартом.** Первый чарт варианта работает как селектор: в режиме редактирования откройте настройки виджета "
              "(значок в правом верхнем углу чарта) → включите **Фильтрация** → **Сохранить**. Щелчок по столбцу/сектору/клетке "
              "фильтрует остальные чарты, повторный щелчок — сбрасывает.",
    "apply":  "**Группа селекторов с кнопками.** Все селекторы варианта — в одном виджете: в окне селектора нажмите **Добавить**, "
              "затем на вкладке **Настройки группы** включите **Кнопка «Применить»** и **Кнопка «Сбросить»**. Учитель сначала "
              "выбирает условия, потом применяет их одним нажатием.",
    "tabs":   "**Две вкладки.** Настройки дашборда → **Вкладки** → **Добавить**: вкладка «Обзор» (индикаторы и чарты варианта) и "
              "вкладка «Для педагогического рассмотрения» (таблица и пояснение). Селектор «Школа» покажите на обеих вкладках "
              "(**Показать на вкладках** → все вкладки).",
}

VARIANTS = {
    1: dict(theme="Учебные затруднения: предшествующие результаты и время самостоятельной подготовки",
            q="У каких сочетаний «ранее не освоенные курсы × время на учёбу» доля G3 ниже 10 выше и как меняются средние оценки по периодам?",
            case="Выявление учебных затруднений и еженедельная оценка продвижения", feature="filter",
            charts=[C("Сводная таблица", "Доля G3 ниже 10 по числу ранее не освоенных курсов", rows="Неосвоенные курсы (группа)", cols="Время на учёбу", m="Доля G3 ниже 10, %"),
                    C("Линейная диаграмма", "Средние оценки по периодам", x="Measure Names", y=["Средняя G1", "Средняя G2", "Средняя G3"], color="Неосвоенные курсы (группа)"),
                    C("Столбчатая диаграмма", "Состав групп по числу ранее не освоенных курсов", x="Неосвоенные курсы (группа)", y=["Учащихся"], color="Категория G3")],
            selectors=[("Время на учёбу", S), ("Категория G3", S)]),
    2: dict(theme="Посещаемость и планирование промежуточной оценки",
            q="Как с ростом пропусков меняются численность группы и доля учащихся с неудовлетворительным итоговым результатом?",
            case="Учёт посещаемости и планирование промежуточной оценки", feature="apply",
            charts=[C("Комбинированная диаграмма", "Посещаемость: численность групп и доля G3 ниже 10", x="Пропуски (группа)",
                      layers=[("Столбчатая", "Учащихся", "Y"), ("Линейная", "Доля G3 ниже 10, %", "Y2")]),
                    C("Точечная диаграмма", "Пропуски и итоговая оценка каждого учащегося", x="Пропуски", y="G3 с разбросом", points="Код", color="Категория G3"),
                    C("Нормированная столбчатая диаграмма", "Доля категорий итогового результата по числу ранее не освоенных курсов", x="Неосвоенные курсы (группа)", y=["Учащихся"], color="Категория G3")],
            selectors=[("Пропуски (группа)", S), ("Неосвоенные курсы (группа)", S)]),
    3: dict(theme="Свободное время и встречи с друзьями: связь с учебными результатами",
            q="Как распределены градации A/B/C/F при разной частоте встреч с друзьями и количестве свободного времени?",
            case="Анализ распределения свободного времени и посещаемости", feature="tabs",
            charts=[C("Нормированная диаграмма с областями", "Структура градаций по частоте встреч с друзьями", x="Встречи с друзьями", y=["Учащихся"], color="Градация"),
                    C("Сводная таблица", "Средняя G3 по свободному времени и встречам с друзьями", rows="Свободное время", cols="Встречи с друзьями", m="Средняя G3"),
                    C("Кольцевая диаграмма", "Градации A/B/C/F", color="Градация", m="Учащихся")],
            selectors=[("Встречи с друзьями", S), ("Градация", S)]),
    4: dict(theme="Поддержка без социальных ярлыков: семья и ресурсы",
            q="Связаны ли отношения в семье, семейная поддержка и интернет дома с итоговой оценкой на уровне групп?",
            case="Поддержка без социальных ярлыков, опора на доступные ресурсы", feature="filter",
            charts=[C("Линейчатая диаграмма", "Доля G3 ниже 10 по качеству отношений в семье", y_dim="Отношения в семье", x_m=["Доля G3 ниже 10, %"]),
                    C("Древовидная диаграмма", "Семейная поддержка и интернет: численность и доля G3 ниже 10", dims=["Семейная поддержка", "Интернет дома"], size="Учащихся", colorm="Доля G3 ниже 10, %"),
                    C("Столбчатая диаграмма", "Средняя G3 по отношениям в семье и поддержке", x="Отношения в семье", y=["Средняя G3"], color="Семейная поддержка", grouped=True)],
            selectors=[("Семейная поддержка", S), ("Интернет дома", S)]),
    5: dict(theme="Дифференцированные задания по текущим достижениям",
            q="Как распределены уровни текущих оценок и куда учащиеся переходят от G1 к G3?",
            case="Дифференцированные задания по текущим достижениям", feature="apply",
            charts=[C("Круговая диаграмма", "Уровни оценки второго периода", color="Уровень G2", m="Учащихся"),
                    C("Нормированная линейчатая диаграмма", "Градации у учащихся со школьной поддержкой и без", y_dim="Школьная поддержка", x_m=["Учащихся"], color="Градация"),
                    C("Сводная таблица", "Матрица переходов: уровень G1 → уровень G3", rows="Уровень G1", cols="Уровень G3", m="Учащихся")],
            selectors=[("Уровень G1", S), ("Платные занятия", S)]),
    6: dict(theme="Мотивационный модуль и предметные консультации",
            q="Отличается ли итоговая оценка у тех, кто планирует высшее образование, при разном времени на учёбу; сколько учащихся выделяет правило?",
            case="Мотивационный модуль и предметные консультации", feature="tabs",
            charts=[C("Столбчатая диаграмма", "Средняя G3 по времени подготовки и намерению продолжить обучение", x="Время на учёбу", y=["Средняя G3"], color="Хочет высшее", grouped=True),
                    C("Кольцевая диаграмма", "Планируют высшее образование", color="Хочет высшее", m="Учащихся"),
                    C("Воронка", "Этапы отбора учащихся по оценкам G1/G2", funnel=["Учащихся", "Выделено правилом", "Подтверждено по G3"])],
            selectors=[("Хочет высшее", S), ("Время на учёбу", S)]),
    7: dict(theme="Планирование времени: дорога, учёба и пропуски",
            q="Как доля G3 ниже 10 различается по времени дороги до школы, времени самостоятельной подготовки и посещаемости в городской и сельской местности?",
            case="Планирование времени и контроль посещаемости", feature="filter",
            charts=[C("Сводная таблица", "Доля G3 ниже 10: время дороги × время на учёбу", rows="Время дороги", cols="Время на учёбу", m="Доля G3 ниже 10, %"),
                    C("Линейная диаграмма", "Доля G3 ниже 10 по группам посещаемости", x="Пропуски (группа)", y=["Доля G3 ниже 10, %"], color="Местность"),
                    C("Линейчатая диаграмма", "Средние пропуски по времени дороги", y_dim="Время дороги", x_m=["Средние пропуски"])],
            selectors=[("Время дороги", S), ("Местность", S)]),
    8: dict(theme="Подбор бесплатных цифровых и очных форм поддержки",
            q="Какие сочетания видов поддержки встречаются чаще и какая у них средняя итоговая оценка?",
            case="Подбор бесплатных цифровых и очных форм поддержки", feature="apply",
            charts=[C("Древовидная диаграмма", "Сочетания поддержки: численность и средняя G3", dims=["Сочетание поддержки"], size="Учащихся", colorm="Средняя G3"),
                    C("Линейчатая диаграмма", "Средняя G3 по сочетаниям поддержки", y_dim="Сочетание поддержки", x_m=["Средняя G3"], sort="Средняя G3"),
                    C("Нормированная столбчатая диаграмма", "Градации и внеурочная активность", x="Внеурочная активность", y=["Учащихся"], color="Градация")],
            selectors=[("Школьная поддержка", S), ("Семейная поддержка", S), ("Платные занятия", S)]),
    9: dict(theme="Индивидуальные учебные цели с учётом предыдущих результатов",
            q="Как связаны G1 и G3 у учащихся с разным числом ранее не освоенных курсов и с какого уровня G1 чаще встречается G3 ниже 10?",
            case="Индивидуальные цели без выводов о способностях по семейным факторам", feature="tabs",
            charts=[C("Точечная диаграмма", "G1 и G3: каждая точка — учащийся", x="G1 с разбросом", y="G3 с разбросом", points="Код", color="Неосвоенные курсы (группа)"),
                    C("Столбчатая диаграмма", "Доля G3 ниже 10 по уровню G1", x="Уровень G1", y=["Доля G3 ниже 10, %"]),
                    C("Сводная таблица", "Численность: неосвоенные курсы × уровень G1", rows="Неосвоенные курсы (группа)", cols="Уровень G1", m="Учащихся")],
            selectors=[("Неосвоенные курсы (группа)", S), ("Уровень G1", S)]),
    10: dict(theme="Режим и регулярность учебной работы",
             q="Как с частотой встреч с друзьями меняются пропуски, итоговая оценка и состав категорий итогового результата?",
             case="Анализ режима и регулярности учебной работы", feature="filter",
             charts=[C("Диаграмма с областями", "Категории G3 по частоте встреч с друзьями (с накоплением)", x="Встречи с друзьями", y=["Учащихся"], color="Категория G3"),
                     C("Комбинированная диаграмма", "Встречи с друзьями: средние пропуски и средняя G3", x="Встречи с друзьями",
                       layers=[("Столбчатая", "Средние пропуски", "Y"), ("Линейная", "Средняя G3", "Y2")]),
                     C("Кольцевая диаграмма", "Градации A/B/C/F", color="Градация", m="Учащихся")],
             selectors=[("Встречи с друзьями", S), ("Пропуски (группа)", S)]),
    11: dict(theme="Планирование педагогической поддержки по итогам первого периода",
             q="При каких значениях G1 чаще встречается G3 ниже 10 и сколько учащихся с G3 ниже 10 находит правило «G1 ниже порога»?",
             case="Планирование педагогической поддержки по итогам первого периода", feature="apply", rule="G1",
             charts=[C("Столбчатая диаграмма", "Доля G3 ниже 10 при каждой оценке G1", x="G1", y=["Доля G3 ниже 10, %"]),
                     C("Воронка", "Этапы отбора учащихся по оценкам G1/G2", funnel=["Учащихся", "Выделено правилом", "Подтверждено по G3"]),
                     C("Нормированная столбчатая диаграмма", "Результат отбора по уровню G1", x="Уровень G1", y=["Учащихся"], color="Результат отбора")],
             selectors=[("Результат отбора", S), ("Неосвоенные курсы (группа)", S)]),
    12: dict(theme="Подбор доступных форм школьной поддержки",
             q="Различается ли структура градаций по причинам выбора школы и как распределена школьная поддержка?",
             case="Подбор доступных форм школьной поддержки", feature="tabs",
             charts=[C("Древовидная диаграмма", "Школа и причина выбора: численность и доля G3 ниже 10", dims=["Школа", "Причина выбора"], size="Учащихся", colorm="Доля G3 ниже 10, %"),
                     C("Нормированная линейчатая диаграмма", "Градации по причине выбора школы", y_dim="Причина выбора", x_m=["Учащихся"], color="Градация"),
                     C("Круговая диаграмма", "Школьная поддержка", color="Школьная поддержка", m="Учащихся")],
             selectors=[("Причина выбора", S), ("Внеурочная активность", S)]),
    13: dict(theme="Обратная связь и учебные ресурсы",
             q="Как средняя итоговая оценка меняется в зависимости от отношений в семье, семейной поддержки и времени на учёбу?",
             case="Подбор форм обратной связи и учебных ресурсов", feature="filter",
             charts=[C("Линейная диаграмма", "Средняя G3 по отношениям в семье", x="Отношения в семье", y=["Средняя G3"], color="Семейная поддержка"),
                     C("Сводная таблица", "Средняя G3: отношения в семье × время на учёбу", rows="Отношения в семье", cols="Время на учёбу", m="Средняя G3"),
                     C("Столбчатая диаграмма", "Уровни G1 и семейная поддержка", x="Уровень G1", y=["Учащихся"], color="Семейная поддержка")],
             selectors=[("Семейная поддержка", S), ("Отношения в семье", S)]),
    14: dict(theme="Учебный режим без медицинских интерпретаций",
             q="При каких сочетаниях свободного времени и встреч с друзьями итоговая оценка выше и как распределены пропуски?",
             case="Работа с учебным режимом без медицинских интерпретаций", feature="apply",
             charts=[C("Сводная таблица", "Средняя G3 по свободному времени и встречам с друзьями", rows="Свободное время", cols="Встречи с друзьями", m="Средняя G3"),
                     C("Нормированная диаграмма с областями", "Структура градаций по свободному времени", x="Свободное время", y=["Учащихся"], color="Градация"),
                     C("Линейчатая диаграмма", "Пропуски и категории итогового результата", y_dim="Пропуски (группа)", x_m=["Учащихся"], color="Категория G3")],
             selectors=[("Свободное время", S), ("Градация", S)]),
    15: dict(theme="Офлайн- и онлайн-ресурсы, задания разного уровня",
             q="Одинаково ли G2 связан с G3 у учащихся с интернетом дома и без него; различаются ли категории итогового результата по планам на высшее?",
             case="Очные и цифровые учебные ресурсы и разноуровневые задания", feature="tabs",
             charts=[C("Точечная диаграмма", "G2 и G3 с учётом интернета дома", x="G2 с разбросом", y="G3 с разбросом", points="Код", color="Интернет дома"),
                     C("Кольцевая диаграмма", "Интернет дома", color="Интернет дома", m="Учащихся"),
                     C("Нормированная столбчатая диаграмма", "Категории G3 и планы на высшее образование", x="Хочет высшее", y=["Учащихся"], color="Категория G3")],
             selectors=[("Интернет дома", S), ("Хочет высшее", S)]),
    16: dict(theme="Консультации и восстановление пропущенных тем",
             q="Как пропуски и ранее не освоенные курсы связаны с итоговой оценкой и составом градаций?",
             case="Консультации и восстановление пропущенных тем", feature="filter",
             charts=[C("Линейчатая диаграмма", "Средняя G3 по группам пропусков", y_dim="Пропуски (группа)", x_m=["Средняя G3"]),
                     C("Комбинированная диаграмма", "Неосвоенные курсы: численность и средние пропуски", x="Неосвоенные курсы (группа)",
                       layers=[("Столбчатая", "Учащихся", "Y"), ("Линейная", "Средние пропуски", "Y2")]),
                     C("Нормированная линейчатая диаграмма", "Градации и школьная поддержка", y_dim="Школьная поддержка", x_m=["Учащихся"], color="Градация")],
             selectors=[("Пропуски (группа)", S), ("Школьная поддержка", S)]),
    17: dict(theme="Реалистичный график и доступность ресурсов",
             q="Связаны ли местность и время дороги с неудовлетворительным итоговым результатом и итоговой оценкой?",
             case="Реалистичный график и доступность ресурсов", feature="apply",
             charts=[C("Сводная таблица", "Доля G3 ниже 10: местность × время дороги", rows="Местность", cols="Время дороги", m="Доля G3 ниже 10, %"),
                     C("Столбчатая диаграмма", "Средняя G3 по времени дороги в городе и селе", x="Время дороги", y=["Средняя G3"], color="Местность", grouped=True),
                     C("Круговая диаграмма", "Город и село", color="Местность", m="Учащихся")],
             selectors=[("Местность", S), ("Интернет дома", S)]),
    18: dict(theme="Индивидуальные задания с учётом текущих результатов обучения",
             q="Как отличаются оценки G1 и G2 у учащихся, выделенных и не выделенных правилом, с учётом итоговой оценки G3?",
             case="Индивидуальные задания с учётом текущих результатов обучения", feature="tabs",
             charts=[C("Точечная диаграмма", "Сопоставление результатов G1 и G2 с итогом отбора", x="G1 с разбросом", y="G2 с разбросом", points="Код", color="Результат отбора"),
                     C("Сводная таблица", "Численность: уровень G1 × уровень G2", rows="Уровень G1", cols="Уровень G2", m="Учащихся"),
                     C("Воронка", "Этапы отбора учащихся по оценкам G1/G2", funnel=["Учащихся", "Выделено правилом", "Подтверждено по G3"])],
             selectors=[("Результат отбора", S), ("Время на учёбу", S)]),
    19: dict(theme="Вид поддержки и учебный профиль",
             q="Как меняются средние оценки по периодам у учащихся с платными занятиями и без; как сочетаются виды поддержки?",
             case="Сопоставление вида поддержки и учебного профиля", feature="filter",
             charts=[C("Линейная диаграмма", "Средние оценки по периодам и платные занятия", x="Measure Names", y=["Средняя G1", "Средняя G2", "Средняя G3"], color="Платные занятия"),
                     C("Древовидная диаграмма", "Школьная → семейная поддержка: численность и средняя G3", dims=["Школьная поддержка", "Семейная поддержка"], size="Учащихся", colorm="Средняя G3"),
                     C("Нормированная столбчатая диаграмма", "Градации и платные занятия", x="Платные занятия", y=["Учащихся"], color="Градация")],
             selectors=[("Платные занятия", S), ("Градация", S)]),
    20: dict(theme="Учебная цель, план недели и короткая обратная связь",
             q="Как внеурочная активность, планы на высшее, время на учёбу и встречи с друзьями связаны с итоговой оценкой?",
             case="Учебная цель, план недели и короткая обратная связь", feature="apply",
             charts=[C("Столбчатая диаграмма", "Средняя G3: внеурочная активность и планы на высшее", x="Внеурочная активность", y=["Средняя G3"], color="Хочет высшее", grouped=True),
                     C("Нормированная диаграмма с областями", "Категории G3 по времени на учёбу", x="Время на учёбу", y=["Учащихся"], color="Категория G3"),
                     C("Линейчатая диаграмма", "Доля G3 ниже 10 по частоте встреч с друзьями", y_dim="Встречи с друзьями", x_m=["Доля G3 ниже 10, %"])],
             selectors=[("Внеурочная активность", S), ("Хочет высшее", S), ("Встречи с друзьями", S)]),
    21: dict(theme="Задания разной сложности и формирующее оценивание",
             q="Сколько учащихся каждой градации при разном времени на учёбу и как с ним связаны отношения в семье?",
             case="Задания разной сложности и формирующее оценивание", feature="tabs",
             charts=[C("Столбчатая диаграмма", "Градации по времени на учёбу (сгруппированные столбцы)", x="Время на учёбу", y=["Учащихся"], color="Градация", grouped=True),
                     C("Сводная таблица", "Средняя G3: отношения в семье × время на учёбу", rows="Отношения в семье", cols="Время на учёбу", m="Средняя G3"),
                     C("Кольцевая диаграмма", "Градации A/B/C/F", color="Градация", m="Учащихся")],
             selectors=[("Градация", S), ("Семейная поддержка", S)]),
    22: dict(theme="Адаптация графика и школьная поддержка",
             q="Как учащиеся распределены по школам и времени дороги и где чаще встречаются итоговые оценки ниже 10?",
             case="Адаптация графика и школьная поддержка", feature="filter",
             charts=[C("Древовидная диаграмма", "Школа и время дороги: численность и доля G3 ниже 10", dims=["Школа", "Время дороги"], size="Учащихся", colorm="Доля G3 ниже 10, %"),
                     C("Комбинированная диаграмма", "Время дороги: численность и средняя G3", x="Время дороги",
                       layers=[("Столбчатая", "Учащихся", "Y"), ("Линейная", "Средняя G3", "Y2")]),
                     C("Нормированная линейчатая диаграмма", "Категории G3 по причине выбора школы", y_dim="Причина выбора", x_m=["Учащихся"], color="Категория G3")],
             selectors=[("Время дороги", S), ("Причина выбора", S)]),
    23: dict(theme="Планирование нагрузки по профилю группы",
             q="Как меняется средняя оценка по периодам при разном числе ранее не освоенных курсов и как связаны свободное время и итог?",
             case="Планирование нагрузки по профилю группы", feature="apply",
             charts=[C("Линейная диаграмма", "Средние оценки по периодам и ранее не освоенные курсы", x="Measure Names", y=["Средняя G1", "Средняя G2", "Средняя G3"], color="Неосвоенные курсы (группа)"),
                     C("Диаграмма с областями", "Градации по свободному времени (с накоплением)", x="Свободное время", y=["Учащихся"], color="Градация"),
                     C("Сводная таблица", "Средняя G3: свободное время × неосвоенные курсы", rows="Свободное время", cols="Неосвоенные курсы (группа)", m="Средняя G3")],
             selectors=[("Неосвоенные курсы (группа)", S), ("Свободное время", S)]),
    24: dict(theme="Сочетание учебных материалов, консультаций и промежуточной оценки",
             q="При каких сочетаниях поддержки доля G3 ниже 10 выше и сколько учащихся с G3 ниже 10 находит правило?",
             case="Сочетание учебных материалов, консультаций и промежуточной оценки", feature="tabs",
             charts=[C("Линейчатая диаграмма", "Доля G3 ниже 10 по сочетаниям поддержки", y_dim="Сочетание поддержки", x_m=["Доля G3 ниже 10, %"], sort="Доля G3 ниже 10, %"),
                     C("Воронка", "Этапы отбора учащихся по оценкам G1/G2", funnel=["Учащихся", "Выделено правилом", "Подтверждено по G3"]),
                     C("Сводная таблица", "Доля G3 ниже 10: школьная × семейная поддержка", rows="Школьная поддержка", cols="Семейная поддержка", m="Доля G3 ниже 10, %")],
             selectors=[("Школьная поддержка", S), ("Семейная поддержка", S), ("Интернет дома", S)]),
    25: dict(theme="Анализ учебных затруднений, индивидуальные задания и педагогическая поддержка",
             q="Как доля итоговых оценок G3 ниже 10 различается по числу ранее не освоенных курсов и частоте встреч с друзьями? Как посещаемость связана с итоговым результатом и какова полнота предварительного отбора по G2?",
             case="Выявление учебных затруднений, подбор заданий и двухнедельная оценка продвижения", feature="filter",
             charts=[C("Сводная таблица", "Доля G3 ниже 10: ранее не освоенные курсы × встречи с друзьями", rows="Неосвоенные курсы (группа)", cols="Встречи с друзьями", m="Доля G3 ниже 10, %"),
                     C("Точечная диаграмма", "Пропуски и итоговая оценка каждого учащегося", x="Пропуски", y="G3 с разбросом", points="Код", color="Категория G3"),
                     C("Комбинированная диаграмма", "Посещаемость: численность групп и доля G3 ниже 10", x="Пропуски (группа)",
                       layers=[("Столбчатая", "Учащихся", "Y"), ("Линейная", "Доля G3 ниже 10, %", "Y2")])],
             selectors=[("Неосвоенные курсы (группа)", S), ("Результат отбора", S), ("Пропуски", INP)]),
}

# ---------------------------------------------------------------------------
# Описание чарта для визарда DataLens и эталонный агрегат из Colab
# ---------------------------------------------------------------------------
def chart_measures(ch):
    out = []
    for k in ("y", "x_m", "funnel"):
        v = ch.get(k, [])
        if isinstance(v, list):                      # у точечной диаграммы y — поле, а не список показателей
            out += v
    for k in ("m", "size", "colorm"):
        if ch.get(k):
            out.append(ch[k])
    out += [m for _, m, _ in ch.get("layers", [])]
    return list(dict.fromkeys(out))

def chart_sections(ch):
    t = ch["t"]
    if t in ("Столбчатая диаграмма", "Нормированная столбчатая диаграмма", "Линейная диаграмма",
             "Диаграмма с областями", "Нормированная диаграмма с областями"):
        s = [("X", ch["x"] + (" (появится сам, перетащите из «Цвета» в X)" if ch["x"] == "Measure Names" else "")),
             ("Y", ", ".join(ch["y"]))]
        if ch.get("color"):
            s.append(("Цвета", ch["color"]))
        if ch.get("grouped"):
            s.append(("Группировка", f"продублируйте «{ch['color']}» в секцию X (рядом с «{ch['x']}») — столбцы встанут рядом"))
        s.append(("Подписи", ch["y"][0]))
    elif t in ("Линейчатая диаграмма", "Нормированная линейчатая диаграмма"):
        s = [("Y", ch["y_dim"]), ("X", ", ".join(ch["x_m"]))]
        if ch.get("color"):
            s.append(("Цвета", ch["color"]))
        if ch.get("sort"):
            s.append(("Сортировка", f"{ch['sort']} (по убыванию)"))
        s.append(("Подписи", ch["x_m"][0]))
    elif t == "Точечная диаграмма":
        s = [("X", ch["x"]), ("Y", ch["y"]), ("Точки", ch["points"]), ("Цвета", ch["color"])]
    elif t in ("Круговая диаграмма", "Кольцевая диаграмма"):
        s = [("Цвет", ch["color"]), ("Показатели", ch["m"]), ("Подписи", ch["m"])]
    elif t == "Древовидная диаграмма":
        s = [("Измерения", " → ".join(ch["dims"])), ("Размер", ch["size"]), ("Цвета", ch["colorm"])]
    elif t == "Сводная таблица":
        ms = ch["m"] if ch["m"] == "Учащихся" else ch["m"] + ", Учащихся (размер клетки n — Measure Names сам встанет в «Столбцы»)"
        s = [("Строки", ch["rows"]), ("Столбцы", ch["cols"]), ("Показатели", ms), ("Цвета", ch["m"] + " (градиент заливки клеток)")]
    elif t == "Воронка":
        s = [("Показатели", " → ".join(ch["funnel"]) + " (Measure Names сам появится в «Категории»)"), ("Подписи", "Measure Values")]
    elif t == "Комбинированная диаграмма":
        s = [("X (общая для слоёв)", ch["x"])]
        for i, (lt, m, ax) in enumerate(ch["layers"], 1):
            s.append((f"Слой {i}: тип «{lt}»", f"{m} → секция {ax}"))
    else:
        raise ValueError(t)
    return s

def chart_dims(ch):
    t = ch["t"]
    if t == "Сводная таблица":
        return [ch["rows"], ch["cols"]]
    if t == "Древовидная диаграмма":
        return list(ch["dims"])
    if t in ("Круговая диаграмма", "Кольцевая диаграмма"):
        return [ch["color"]]
    if t == "Воронка":
        return []
    if t == "Точечная диаграмма":
        return None                                  # точка — учащийся, агрегата нет
    if t.startswith("Линейчат") or t.startswith("Нормированная линейчат"):
        return [ch["y_dim"]] + ([ch["color"]] if ch.get("color") else [])
    dims = [] if ch.get("x") == "Measure Names" else [ch["x"]]
    if ch.get("color"):
        dims.append(ch["color"])
    return dims

def agg_table(d, dims, measures):
    if not dims:
        return pd.DataFrame([{m: MEASURES[m][1](d) for m in measures}], index=["все учащиеся"])
    rows = []
    for key, g in d.groupby(dims, observed=True):
        key = key if isinstance(key, tuple) else (key,)
        rows.append({**dict(zip(dims, key)), **{m: MEASURES[m][1](g) for m in measures}})
    return pd.DataFrame(rows).set_index(dims)

def expected(d, ch):
    """Эталонный агрегат: то, что должен показать чарт DataLens при выборе «все значения» в селекторах."""
    dims, ms = chart_dims(ch), chart_measures(ch)
    if dims is None:                                 # точечная: проверяем число точек и средние осей
        return pd.DataFrame({"точек (учащихся)": [len(d)],
                             f"среднее {ch['x']}": [round(d[ch["x"]].mean(), 2)],
                             f"среднее {ch['y']}": [round(d[ch["y"]].mean(), 2)]}, index=["все учащиеся"])
    t = agg_table(d, dims, ms if ms else ["Учащихся"])
    if ch["t"] != "Сводная таблица" and "Учащихся" not in t.columns and dims:
        t["n"] = agg_table(d, dims, ["Учащихся"])["Учащихся"]
    if ch["t"] == "Сводная таблица":
        if ch["m"] == "Учащихся":
            return t["Учащихся"].unstack(ch["cols"]).fillna(0).astype(int)
        n = agg_table(d, dims, ["Учащихся"])["Учащихся"]
        cell = t[ch["m"]].map(lambda v: f"{v:g}") + " (n=" + n.astype(str) + ")"
        return cell.unstack(ch["cols"]).fillna("—")
    return t

print("Служебная ячейка выполнена: вариантов —", len(VARIANTS), "· показателей —", len(MEASURES))


GRADE_BINS = [-1, 9, 13, 15, 20]
YESNO = {"yes": "да", "no": "нет"}

def build_base(raw):
    """Таблица для DataLens: одна строка — один учащийся, подписи по-русски, только нужные поля."""
    d = pd.DataFrame()
    d["Код"] = [f"S{i:03d}" for i in range(1, len(raw) + 1)]
    d["Школа"] = raw["school"].values
    d["Местность"] = raw["address"].map({"U": "город", "R": "село"}).values
    d["Причина выбора"] = raw["reason"].map({"home": "рядом с домом", "reputation": "репутация школы",
                                             "course": "учебная программа", "other": "другое"}).values
    d["Время дороги"] = raw["traveltime"].map({1: "1 · до 15 мин", 2: "2 · 15–30 мин", 3: "3 · 30–60 мин", 4: "4 · более 60 мин"}).values
    d["Время на учёбу"] = raw["studytime"].map({1: "1 · менее 2 ч", 2: "2 · 2–5 ч", 3: "3 · 5–10 ч", 4: "4 · более 10 ч"}).values
    d["Неосвоенные курсы"] = raw["failures"].values
    d["Неосвоенные курсы (группа)"] = np.where(raw["failures"] >= 2, "2+", raw["failures"].astype(str))
    d["Пропуски"] = raw["absences"].values
    d["Пропуски (группа)"] = pd.cut(raw["absences"], [-1, 0, 5, 10, 1000],
                                    labels=["1 · нет", "2 · 1–5", "3 · 6–10", "4 · 11 и более"]).astype(str).values
    d["Встречи с друзьями"] = raw["goout"].values
    d["Свободное время"] = raw["freetime"].values
    d["Отношения в семье"] = raw["famrel"].values
    for src, name in [("higher", "Хочет высшее"), ("schoolsup", "Школьная поддержка"), ("famsup", "Семейная поддержка"),
                      ("paid", "Платные занятия"), ("internet", "Интернет дома"), ("activities", "Внеурочная активность")]:
        d[name] = raw[src].map(YESNO).values
    parts = np.column_stack([np.where(raw["schoolsup"] == "yes", "школа", ""), np.where(raw["famsup"] == "yes", "семья", ""),
                             np.where(raw["paid"] == "yes", "платные", "")])
    d["Сочетание поддержки"] = [" + ".join(p for p in row if p) or "не отмечены перечисленные формы" for row in parts]
    rng = np.random.default_rng(42)
    for g in ["G1", "G2", "G3"]:
        d[g] = raw[g].values
    for g in ["G1", "G2", "G3"]:
        d[f"{g} с разбросом"] = (raw[g].values + rng.uniform(-0.3, 0.3, len(raw))).round(2)
    for g in ["G1", "G2", "G3"]:
        d[f"Уровень {g}"] = pd.cut(raw[g], GRADE_BINS, labels=["F (0–9)", "C (10–13)", "B (14–15)", "A (16–20)"]).astype(str).values
    d["Градация"] = pd.cut(raw["G3"], GRADE_BINS, labels=["F", "C", "B", "A"]).astype(str).values
    d["G3 ниже 10"] = (raw["G3"] < 10).astype(int).values
    d["Категория G3"] = np.where(d["G3 ниже 10"] == 1, "G3 < 10", "G3 ≥ 10")
    d["Динамика G3-G1"] = (raw["G3"] - raw["G1"]).values
    d["Динамика"] = np.select([d["Динамика G3-G1"] > 0, d["Динамика G3-G1"] < 0], ["выросла", "снизилась"], "без изменений")
    # Только индикатор необычного сочетания оценок: он НЕ подтверждает отсутствие на аттестации.
    d["G3=0 при G1>0"] = ((raw["G3"] == 0) & (raw["G1"] > 0)).astype(int).values
    return d

def add_signals(d, rule_col, t):
    """Пороговый отбор по G1/G2 для последующей проверки по G3 (TP/FP/FN/TN) и подготовки рекомендаций."""
    d = d.copy()
    d["Выделен правилом"] = (d[rule_col] < t).astype(int)
    d["TP"] = ((d["Выделен правилом"] == 1) & (d["G3 ниже 10"] == 1)).astype(int)
    d["FP"] = ((d["Выделен правилом"] == 1) & (d["G3 ниже 10"] == 0)).astype(int)
    d["FN"] = ((d["Выделен правилом"] == 0) & (d["G3 ниже 10"] == 1)).astype(int)
    d["TN"] = ((d["Выделен правилом"] == 0) & (d["G3 ниже 10"] == 0)).astype(int)
    d["Результат отбора"] = np.select([d["TP"] == 1, d["FP"] == 1, d["FN"] == 1],
                                    ["1 · выделен, G3 < 10", "2 · выделен, G3 ≥ 10", "3 · не выделен, G3 < 10"], "4 · не выделен, G3 ≥ 10")
    order = d[d["Выделен правилом"] == 1].sort_values([rule_col, "Пропуски"], ascending=[True, False]).index
    d["Очередность рассмотрения"] = 0
    d.loc[order, "Очередность рассмотрения"] = range(1, len(order) + 1)

    def plan(r):
        a = []
        if r[rule_col] == 0:
            return "уточнить причины нулевой оценки за период · решение педагога"
        if r["G1"] < 10 or r["G2"] < 10: a.append("выявить учебные затруднения и подобрать 2–3 задания")
        if r["Неосвоенные курсы"] > 0:            a.append("промежуточная проверка освоения материала через неделю")
        if r["Пропуски"] >= 10:         a.append("план освоения материала пропущенных занятий")
        if r["Школьная поддержка"] == "нет": a.append("обсудить возможность дополнительной школьной консультации")
        a.append("обсудить рекомендации с учащимся и пересмотреть через 14 дней")
        return " · ".join(a)
    d["Возможные формы поддержки"] = d.apply(plan, axis=1)
    return d
````

````python
import io, os, re, urllib.request, warnings, zipfile
warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None); pd.set_option("display.width", 160)

URLS = [
    "https://raw.githubusercontent.com/BosenkoTM/Big-data-analysis-methods/cbfd565cab7dae125042b9678258793bb971eaec/2026/lw/lw_02/student-mat.csv",  # зафиксированная версия набора из репозитория курса
    "https://raw.githubusercontent.com/sachanganesh/student-performance-prediction/refs/heads/master/data/student-mat.csv",
    "https://archive.ics.uci.edu/static/public/320/student+performance.zip",                                                  # официальный архив UCI
]

def load_student_mat(urls):
    if os.path.exists("student-mat.csv"):                  # локальная копия рядом с ноутбуком
        return pd.read_csv("student-mat.csv", sep=";"), "локальный файл student-mat.csv"
    errors = []
    for url in urls:
        try:
            with urllib.request.urlopen(url, timeout=30) as r:
                raw = r.read()
            if url.endswith(".zip"):                       # в архиве UCI вложен student.zip
                outer = zipfile.ZipFile(io.BytesIO(raw))
                raw = zipfile.ZipFile(io.BytesIO(outer.read("student.zip"))).read("student-mat.csv")
            frame = pd.read_csv(io.BytesIO(raw), sep=";")
            assert {"G1", "G2", "G3", "studytime"} <= set(frame.columns), "нет нужных столбцов"
            return frame, url
        except Exception as exc:
            errors.append(f"{url} → {type(exc).__name__}: {exc}")
    raise RuntimeError("Не удалось загрузить данные:\n" + "\n".join(errors))

def check_filled(where, values):
    # Проверяет, что вместо «...» вписано значение; если нет — называет незаполненную строку.
    empty = [name for name, v in values.items() if v is ...]
    if empty:
        raise ValueError(f"Не заполнено ({where}): " + ", ".join(empty)
                         + ". Замените «...» на значение из подсказки «← впишите:» и запустите ячейку снова.")

df_raw, USED_URL = load_student_mat(URLS)
# Контроль воспроизводимости: не продолжаем расчёт на другом файле или версии набора.
assert df_raw.shape == (395, 33), f"Ожидается UCI student-mat.csv (395 × 33); получено {df_raw.shape}"
assert {"school", "G1", "G2", "G3", "absences", "failures", "goout"} <= set(df_raw.columns), "Не найдены обязательные поля набора UCI"
assert int((df_raw["G3"] < 10).sum()) == 130 and int((df_raw["G2"] < 10).sum()) == 146, (
    "Контрольные численности не совпали с учебным набором UCI student-mat.csv; проверьте источник данных"
)
print("Источник:", USED_URL)
print("Размер:", df_raw.shape, "— учащихся и признаков")
df_raw.head(3)
````

# Шаг 2. Проверка качества и поля для DataLens

**Ячейка готова — просто запустите.** Проверяем пропуски, дубликаты и диапазоны оценок, затем функция `build_base` собирает таблицу для DataLens: **одна строка — один учащийся**, подписи по-русски, только нужные поля.

Обратите внимание на три приёма, которые экономят время в DataLens:

* **Порядок категорий задан префиксом** («1 · до 15 мин», «2 · 15–30 мин» …): DataLens сортирует подписи по алфавиту, и без номера «более 60 мин» встал бы раньше «до 15 мин».
* **Числовые признаки разбиты на группы заранее** («Пропуски (группа)», «Неосвоенные курсы (группа)», «Уровень G1»): не нужно писать формулы группировки в DataLens.
* **Признаки 0/1** («G3 ниже 10», «Выделен правилом», «TP», «FN» …): любой процент в DataLens считается простой формулой `AVG` или `SUM`.

````python
print("Пропусков всего:", int(df_raw.isna().sum().sum()))
print("Полных дубликатов:", int(df_raw.duplicated().sum()))
grades = ["G1", "G2", "G3"]
display(pd.DataFrame({"min": df_raw[grades].min(), "max": df_raw[grades].max(), "среднее": df_raw[grades].mean().round(2),
                      "вне 0–20": ((df_raw[grades] < 0) | (df_raw[grades] > 20)).sum(), "нулей": (df_raw[grades] == 0).sum()}))

base = build_base(df_raw.drop_duplicates().reset_index(drop=True))
print(f"Таблица для DataLens: {base.shape[0]} строк × {base.shape[1]} столбцов (без выделений по правилу — они появятся в шаге 3)")
display(base["Категория G3"].value_counts().to_frame("учащихся").assign(доля=lambda t: (t["учащихся"] / len(base)).round(3)))
print("Записей с G3 = 0 при G1 > 0 (причина неизвестна):", int(base["G3=0 при G1>0"].sum()))
base.head(3)
````

**Как интерпретировать результаты.** Отсутствие пропусков значений и дубликатов подтверждается проверкой в коде, а не предполагается заранее. Нулевая итоговая оценка `G3 = 0` при `G1 > 0` **не доказывает неявку на аттестацию**: причина нулевой оценки из этих данных неизвестна.
